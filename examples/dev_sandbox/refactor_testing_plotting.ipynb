```python
import clearwater_riverine as cwr
from pathlib import Path
import pandas as pd
```

```python
# read config
working_dir = Path.cwd()
notebook_path = working_dir.parent.parent
network_path = notebook_path / 'examples/data_temp/sumwere_creek_coarse_p48'
config_file = network_path / 'example_config_refactor_no_chunk.yml'
```

```python
# instantiate model
transport_model = cwr.ClearwaterRiverine(
    config_filepath=config_file,
)
```

```python
# run model
transport_model.run()
```

```python
transport_model.plot(constituent_name='water_temperature')
```

```python
transport_model.static_plot(constituent_name='water_temperature', plotting_datetime='2022-05-13 08:28:00', clim=(12, 20))
```

```python
transport_model.quick_plot(constituent_name='water_temperature')
```

```python
transport_model.registry.get('node_x')
```

```python
# you can also call the plotter outside of Riverine like this:
from clearwater_riverine.plotting import RiverinePlotter

# give it a registry and a crs
plotter = RiverinePlotter(registry=transport_model.registry, crs='EPSG:26916')
```

```python
# call it to plot
plotter.dynamic_plot(constituent_name = 'water_temperature')
```
