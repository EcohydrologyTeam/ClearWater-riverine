# Run Riverine

## Imports

```python
import clearwater_riverine as cwr
from pathlib import Path
import pandas as pd
```

## Paths

```python
# read config
working_dir = Path.cwd()
notebook_path = working_dir.parent.parent
network_path = notebook_path / 'examples/data_temp/sumwere_creek_coarse_p48'
config_file = network_path / 'example_config_refactor_no_chunk.yml'
```

## Initialize and Run Model

```python
# instantiate model
transport_model = cwr.ClearwaterRiverine(
    config_filepath=config_file,
)
```

```python
# run model
transport_model.run()
```

## Post-Processing

Run for the tracer, then get the "answer" using 100 everywhere. 

```python
# post-processing: modeled results
mass_balance_model = transport_model.calculate_mass_balance(constituent_name='tracer')

# answer: 100 everywhere
mass_balance_answer = transport_model.calculate_mass_balance(
    constituent_name='tracer',
    calculate_answer=True,
    answer_value=100
)
```

Look at global results:

```python
combined = pd.concat(
    [mass_balance_model['global'], mass_balance_answer['global']],
    keys=['mass_balance_model', 'mass_balance_answer']
)
combined
```
